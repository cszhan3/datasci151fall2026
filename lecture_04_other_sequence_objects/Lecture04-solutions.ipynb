#### **7)** Using NumPy arrays, create a scatter plot of the function

$$q(x) = \sin(\pi x) + 1$$

#### at a set of points in the interval $0 \leq x \leq 2$. (You are free to choose which points are actually used)

```python
x_vals = np.linspace(0, 2, 25)

q_vals = np.sin(np.pi * x_vals) + 1

plt.scatter(x_vals, q_vals)
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.title("Graph of $y = q(x) = \\sin(\\pi x) + 1$")
plt.show()
```

#### **8)** Using NumPy arrays, create a scatter plot of the function $f(g(t))$, where

$$ f(x) = x^2 - 5x + 2 $$

$$ g(t) = \sin(t + 3) $$

#### at a set of points in the interval $2\pi \leq t \leq 4\pi$. (You are free to choose which points are actually used)

<font color = "red">

Note: <font color = "black">  There was a typo in the original question. The interval should be 
$2\pi \leq t \leq 4\pi$ instead of "$x$"

```python
t_vals = np.linspace(2, 4, 25) * np.pi

x_vals = np.sin(t_vals + 3)

f_vals = x_vals**2 - 5*x_vals + 2

plt.scatter(t_vals, f_vals)
plt.xlabel("$t$")
plt.ylabel("$y$")
plt.title("Graph of $y = f(g(t))$")
plt.show()
```

### **9)** Compute some simple statistics for the following set of numbers:

$$ 4.2,\  1.2,\ 11.45,\ -2,\ 0,\ 9,\ 3.3,\ 2\pi $$

<font size = "4">

In particular, compute:

- the mean

- the standard deviation

- the median

- the maximum

- the minimum

Use the built-in NumPy functions: `numpy.mean`, `numpy.std`, `numpy.median`, `numpy.max`, `numpy.min`

```python
nums = np.array([4.2, 1.2, 11.45, -2, 0, 9, 3.3, 2*np.pi])

print("Mean:", np.mean(nums))
print("Std. dev.:", np.std(nums))
print("Median:", np.median(nums))
print("Max:", np.max(nums))
print("Min:", np.min(nums))
```
